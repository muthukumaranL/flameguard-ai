## Object geometry - why we kept 640px inputs

A large fraction of objects are small. Downscaling for speed would push distant flames below the detector's smallest stride.

```python
for f in ['03_box_geometry.png', '05_size_categories.png', '04_center_heatmap.png']:
    display(Image(str(paths.EDA_OUTPUT_DIR / f)))
```

The centre heatmap shows smoke sitting high in the frame and fire low - smoke rises. This is why **vertical flip augmentation is disabled**: an upside-down plume is not a scene that exists.
